```python
#optimizar una búsqueda de cadenas en otra cadena
my_string = """Despertar con el sol cara me llena de alegría mientras disfruto un delicioso desayuno con mi familia.  Logrando una meta laboral que me llena de orgullo. Sin embargo, la tarde se presenta con un tráfico insoportable, 
seguido de una larga y tediosa reunión. Me duele la cabeza, opacando un poco el ánimo.

Una sorpresiva noticia sobre un amigo me pone triste. Un error inesperado en un proyecto laboral me genera frustración, 
y una discusión con un compañero de trabajo aumenta la tensión del día.

A pesar de los altibajos, la reflexión me lleva a aceptar que cada día trae consigo una mezcla de emociones,  algunas agradables y otras no tanto. Lo importante es aprender a manejarlas todas, 
aceptándolas como parte de la vida y sacando lo mejor de cada una."""

positive_words = ["alegría", "disfruto", "Logrando", "relfexión", "agradables"]
negative_words = ["insoportable", "tediosa", "duele", "triste", "fristración" ]
neutral_words = ["larga", "sorpresiva", "inesperado", "tensión"]
score = 0

for positive_word in positive_words:
 score = score + my_string.count(positive_word)
for negative_word in negative_words:
 score = score - my_string.count(negative_word)
for neutral_word in neutral_words:
    score == 0

if score == 0:
 result = "Neutro"
if score > 0:
 result = "Positivo"
if score < 0:
 result = "Negativo"

print("Analizador de sentimientos\n")
print("Comentarios: {}\nscore: {} \nresultado: {}".format(my_string, score, result))

#len
print("La cantidad de caracteres en tu estado son: ", len(my_string))


#split
linea1, linea2, linea3, linea4, linea5, linea6, linea7,linea8 = my_string.split("\n")
print("Párrafos que escribiste: ")
print(linea1)
print(linea2)
print(linea3)
print(linea4)
print(linea5)
print(linea6)
print(linea7)
print(linea8)
```
