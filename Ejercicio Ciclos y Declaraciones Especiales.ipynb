```python
import time 

def countdown(time_sec):    
    while time_sec:        
        mins, secs = divmod(time_sec, 60)        
        timeformat = '{:00d}:{:00d}'.format(mins, secs)        
        print(timeformat, end='\r')        
        time.sleep(1)        
        time_sec -= 1    
        
    print("stop") 
countdown(90)
```

```python
import time 

def countup(time_sec):    
    while time_sec:        
        mins, secs = divmod(time_sec, 60)        
        timeformat = '{:02d}:{:02d}'.format(mins, secs)        
        print(timeformat, end='\r')        
        time.sleep(1)        
        time_sec += 1    
        
    print("stop") 
countup(1)
```
