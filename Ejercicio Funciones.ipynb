```python
def calcular_total_con_tax(productos, precios, tax):
    def aplicar_tax(precio, tax):
        return precio + tax

    print("Menú:")
    for i in range(len(productos)):
        print("{}. {}. ${}".format(i+1, productos[i], precios[i]))

    opcion_usuario = int(input("Ingrese el número del producto que desea (1-{}): ".format(len(productos))))

    
    if 1 <= opcion_usuario <= len(precios):
        precio_seleccionado = precios[opcion_usuario-1]  
        precio_final = aplicar_tax(precio_seleccionado, tax)
        print("El total de {} es: ${}".format(productos[opcion_usuario-1], precio_final))

    else:
        print("Opción inválida.")

productos = ["Pizza", "Hamburguesa", "Refresco", "Papas"]
precios = [15, 10, 3, 7]
tax = 5  

calcular_total_con_tax(productos, precios, tax)
```
