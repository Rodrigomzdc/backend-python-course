```python
my_list = [1,2,3,4,5]
print(my_list)
```

## Conversión de listas

### Cadenas a listas

```python
# Separar palabras
my_string = "curso de backend Python"
my_list = my_string.split()
print(my_string)
print(my_list)
```

```python
# Separar caracteres
my_string = "curso de backend Python"
my_list = list(my_string)
print(my_string)
print(my_list)
```

```python
# Separar los caracteres en una lista de listas
my_string = "curso de backend Python"
my_list = list(map(list, my_string))
print(my_string)
print(my_list)
```

### Listas a Cadenas

```python
my_list = ["curso", "de", "backend", "Python"]
my_string = " ".join(my_list)
print(my_list)
print(my_string)
```

```python
my_list = ['c', 'u', 'r', 's', 'o', ' ', 'd', 'e', ' ', 'b', 'a', 'c', 'k', 'e', 'n', 'd', ' ', 'P', 'y', 't', 'h', 'o', 'n']
my_string = "".join(my_list)
print(my_list)
print(my_string)
```

```python
my_list = [['c'], ['u'], ['r'], ['s'], ['o'], [' '], ['d'], ['e'], [' '], ['b'], ['a'], ['c'], ['k'], ['e'], ['n'], ['d'], [' '], ['P'], ['y'], ['t'], ['h'], ['o'], ['n']]
my_string = "".join(["".join(element)for element in my_list])
print(my_list)
print(my_string)
```

# Métodos para manipulación de listas

## append

Agregar un elemento al final de una lista

```python
my_list = ["Caballo", "Abeja", "Jirada", "Ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.append("Delfín")
print(my_list)
print(len(my_list))
```

```python
my_list.append("Gato")
print(my_list)
print(len(my_list))
```

```python
my_list = [1,2,3,4,5]
print(my_list)
print(len(my_list))
```

```python
my_list.append(6)
print(my_list)
print(len(my_list))
```

```python
my_list.append(300)
print(my_list)
print(len(my_list))
```

## insert

Agregar un valor a una lista en una posición específica

insert(índice, valor)

```python
my_list = ["Caballo", "Abeja", "Jirada", "Ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.insert(2, "Delfín")
print(my_list)
print(len(my_list))
```

```python
my_list.insert(1, "Gato")
print(my_list)
print(len(my_list))
```

## clear

Eliminar todos los elementos de la lista

```python
my_list = ["Caballo", "Abeja", "Jirada", "Ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.clear()
print(my_list)
print(len(my_list))
```

```python
my_list = [1,2,3,4,5]
print(my_list)
print(len(my_list))
```

```python
my_list.clear()
print(my_list)
print(len(my_list))
```

## copy

Copiar los elementos de una lista

```python
my_list = ["Caballo", "Abeja", "Jirada", "Ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list_copy= my_list.copy()
print(my_list_copy)
print(len(my_list_copy))
```

```python
my_list_copy.clear()
print(my_list_copy)
print(len(my_list_copy))
print(my_list)
print(len(my_list))
```

```python
my_list_assigned = my_list
my_list_assigned.clear()
print(my_list_assigned)
print(len(my_list_assigned))
print(my_list)
print(len(my_list))
```

Cuenta cuántas veces aparece un elemento en una lista

```python
print("El número 100 aparece {} veces en la lista".format(my_list.count(300)))
```

## pop

Eliminar elementos de una lista basado en los índices

```python
my_list = [100, 300, 500, 100, 200, 100, 300]
print(my_list)
print(len(my_list))
```

```python
# elimina índice -1 por default
my_list.pop()
print(my_list)
print(len(my_list))
```

```python
my_list.pop(2)
print(my_list)
print(len(my_list))
```
