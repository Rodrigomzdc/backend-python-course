# Variables

### ¿Qué es una variable?
- Son un espacio temporal para guardar un valor con un nombre específico 
- No hay nececidad de declarar el tipo de dato explícitamente 
- Python es un lenguaje tipado dinámicamente, lo que quiere decir que el tipo de dato se define mientras corre el programa
- Las variables en python funcionan diferente que en otros lenguajes de programación
- No necesitamos crear las variables, sólo asignamos un nombre a un valor usando el operador de asignación

### Tipos de Variables
- Variables globales: aquellas que se utilizarán para todo el programa
- Variables locales: aquellas que se utilizarán sólo en un bloque específico 


# Variables Globales
- Las declaramos afuera de funciones
- El valor de la variable puede ser constante para todo el programa
- Cualquier función podrá acceder a estas variables
- Podemos decir que las Variables globales tienen un alcance que es global
- Normalmente son más usadas para constantes
- Mantienen sus valores a través de la ejecución completa de tu programa
- Puede ser accesada por cualquier función 

```python
# Variable Global
my_global_variable = 50

def my_function_1():
    pass
def my_function_2():
    pass
```

```python
#Variables Locales
def my_function_1():
    my_local_variable=5
    
def my_function_2():
    my_local_variable=35
```

```python
# Variable Global
my_global_variable = 50

def my_function_1():
    print("my_function_1 - el valor de my_global_variable es: ", my_global_variable)
    
def my_function_2():
    print("my_function_2 - el valor de my_global_variable es: ", my_global_variable)
```

```python
my_function_1()
```

```python
my_function_2()
```

```python
# Preferencia al valor local, y después al global
my_variable = 50 #sabemos que es global porque está fuera

def my_function_1():
    my_variable = 10 #local
    print("my_function_1 - el valor de my_global_variable es: ", my_variable)
    
def my_function_2():
    my_variable = 20 #local
    print("my_function_2 - el valor de my_global_variable es: ", my_variable)
    
def my_function_3():
    print("my_function_2 - el valor de my_global_variable es: ", my_variable)
```

- Podemos tener una restricción al acceder a una variable local desde otra función

- Lo que dure la función será lo que dure el alcance de nuestra variable

## Plabra Reservada "global"

```python
def my_function_1():
    my_variable = 10 #local
    print("my_function_1 - el valor de my_global_variable es: ", my_variable)
    
def my_function_2():
    global my_variable #Global
    my_variable = 20 #local
    print("my_function_2 - el valor de my_global_variable es: ", my_variable)
    
def my_function_3():
    print("my_function_2 - el valor de my_global_variable es: ", my_variable)
```

```python
my_function_1()
```

```python
my_function_2()
```

```python
my_function_3()
```

```python
def my_function_1():
    my_variable = 10 #local
    print("my_function_1 - el valor de my_global_variable es: ", my_variable)
    
def my_function_2():
    global a, b, c, d, e #Global
    a=10
    b=20
    c=30
    d=40
    e=50
    print("my_function_2 - el valor de a es: ", a)
    print("my_function_2 - el valor de b es: ", b)
    print("my_function_2 - el valor de c es: ", c)
    print("my_function_2 - el valor de d es: ", d)
    print("my_function_2 - el valor de e es: ", e)
    
def my_function_3():
    print("my_function_2 - el valor de a es: ", a)
    print("my_function_2 - el valor de b es: ", b)
    print("my_function_2 - el valor de c es: ", c)
    print("my_function_2 - el valor de d es: ", d)
    print("my_function_2 - el valor de e es: ", e)
```

```python
my_function_2()
```

```python
my_function_3()
```

## Binaria
- Código organizado en Bytes
- Lenguaje Binario
- Lenguaje de las máquinas (1 y 0)

### Valor Base: 2
- Solamente acepta dos dígitos (1 y 0)
- Usamos el prefijo '0b o '0B'

```python
my_decimal_number= 11
print("El valor de my_decimal_numbre: ", my_decimal_number)
```

```python
my_binary_number= 0b11 # 1*2**1 + 1*2**0
print("El valor de my_binary_numbre: ", my_binary_number)
```

```python
my_decimal_number= 111
print("El valor de my_decimal_numbre: ", my_decimal_number)
```

```python
my_binary_number= 0B111 # 1*2**2 + 1*2**1 + 1*2**0
print("El valor de my_binary_numbre: ", my_binary_number)
```

## Octal
### Valor base es 8
- Solamente acepta dígitos del 0 al 7
- Usamos el prefijo '0o o 0O'

```python
my_decimal_number= 17
print("El valor de my_decimal_numbre: ", my_decimal_number)
```

```python
my_octal_number= 0o17 # 1*8**1 + 1*8**0
print("El valor de my_octal_number: ", my_octal_number)
```

```python
my_decimal_number= 153
print("El valor de my_decimal_numbre: ", my_decimal_number)
```

```python
my_octal_number= 0O153 # 1*8**2 + 5*8**1 + 3*8**1
print("El valor de my_octal_number: ", my_octal_number)
```
