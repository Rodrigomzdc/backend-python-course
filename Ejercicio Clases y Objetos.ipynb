```python
class Producto:
    def __init__(self, nombre, precio, descripcion):
        self.nombre = nombre
        self.precio = precio
        self.descripcion = descripcion
        
    def print_attributes(self):
        print("Nombre: ", self.nombre)
        print("Precio: ", self.precio)
        print("Descripción: ", self.descripcion)

    def aplicar_descuento(self, descuento):
        self.precio *= (1 - descuento/100)

class categoria(Producto):
    def __init__(self, nombre, precio, descripcion, tax):
        #Inicializar
        Producto.__init__(self, nombre, precio, descripcion)
        self.tax = 10
    def disponible(self):
        return True
```

```python
Hamburguesa = categoria("Bigmac", 20, "Deliciosa", 10)
Hamburguesa.print_attributes()

print(Hamburguesa.nombre)
print(Hamburguesa.precio)
print(Hamburguesa.tax)
print(Hamburguesa.disponible())

Hamburguesa.aplicar_descuento(10)
print("El precio del producto con descuento", Hamburguesa.precio)
```
